```python
from PIL import Image
import os
from tqdm import tqdm
import random
import multiprocessing
```

```python
def resize_and_overwrite(img_path, size=(128, 128)):
    """Resizes an image and overwrites the original."""
    try:
        img = Image.open(img_path).convert("RGB")  # Convert to RGB
        img = img.resize(size, Image.LANCZOS)  # High-quality resizing
        img.save(img_path)  # Overwrite the original image
    except Exception as e:
        print(f"Error processing {img_path}: {e}")
```

```python
# Multi core
def resize_images(root_dir, size=(128, 128), num_workers=10):
    """Finds all images and resizes them in parallel."""
    if num_workers is None:
        num_workers = multiprocessing.cpu_count()  # Use all available CPU cores

    # Collect all image paths
    img_paths = []
    for subdir, _, files in os.walk(root_dir):  # Recursively go through all subdirectories
        for file in files:
            if file.lower().endswith(".jpeg"):  # Ensure case-insensitive matching
                img_paths.append(os.path.join(subdir, file))

    # Process images in parallel
    with multiprocessing.Pool(processes=num_workers) as pool:
        list(tqdm(pool.imap_unordered(lambda img: resize_and_overwrite(img, size), img_paths), 
                  total=len(img_paths), desc="Resizing images"))
```

```python
# Single core
def resize_images(root_dir, size=(128, 128)):
    for subdir, _, files in os.walk(root_dir):  # Recursively go through all subdirectories
        for file in tqdm(files, desc=f"Processing {subdir}"):
            if file.lower().endswith(".jpeg"):  # Ensure case-insensitive matching
                img_path = os.path.join(root_dir, file)
                img = Image.open(img_path).convert("RGB")  # Convert to RGB if not already
                img = img.resize(size, Image.LANCZOS)  # High-quality resizing
                img.save(img_path)  # Overwrite the original image"
```

```python
def print_paths(root_dir, n=10):
    """Prints paths and sizes of n random images from root_dir and its subfolders."""
    
    # Collect all image paths from subdirectories
    img_paths = []
    for subdir, _, files in os.walk(root_dir):  # Recursively traverse subdirectories
        for file in files:
            if file.lower().endswith(".jpeg"):  # Case-insensitive check for JPEG files
                img_paths.append(os.path.join(subdir, file))

    # Randomly sample n images
    random_files = random.sample(img_paths, min(n, len(img_paths)))
    
    # Print paths and image sizes
    for img_path in random_files:
        try:
            img = Image.open(img_path).convert("RGB")
            print(img_path, img.size)
        except Exception as e:
            print(f"Error processing {img_path}: {e}")
```
