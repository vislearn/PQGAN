# Older lightining version

```python
import torch
from lightning.pytorch.utilities.upgrade_checkpoint import _upgrade
from pytorch_lightning.callbacks.model_checkpoint import ModelCheckpoint
from torch.serialization import add_safe_globals
```

```python
# Allow the ModelCheckpoint class
add_safe_globals([ModelCheckpoint])

# Path to the checkpoint
checkpoint_path = "/path/to/downloaded_models/models/ldm/ldm_kl_F4_Z3_S1/version_0/checkpoints/latest-step=0/ldm_kl_F4_Z3_S1.ckpt"
```

```python
# Upgrade the checkpoint
_upgrade(type("Args", (object,), {"path": checkpoint_path, "map_to_cpu": False, "extension": ""}))
```

# .pth to lightning

```python
import torch
from argparse import Namespace

# Path to the original .pth file
original_pth_path = "/path/to/additional_evaluations/downloaded_models/models/vq_gan_lc/vqgan-lc-100K-f8-dim4.pth"

# Path to save the new Lightning checkpoint
lightning_ckpt_path = "/path/to/additional_evaluations/downloaded_models/models/vq_gan_lc/vqgan-lc-100k-f8-dim4-lightning.ckpt"

# Allowlist argparse.Namespace as a safe global
torch.serialization.add_safe_globals([Namespace])

# Load the .pth file
checkpoint = torch.load(original_pth_path, map_location="cpu")

# Print the keys in the checkpoint
print("Checkpoint keys:", checkpoint.keys())

# Extract the state dictionary (adjust the key if necessary)
state_dict = checkpoint.get("state_dict", checkpoint)  # Use "state_dict" if it exists, otherwise use the whole checkpoint

# Create a new Lightning-compatible checkpoint
lightning_checkpoint = {
    "state_dict": state_dict,
    "pytorch-lightning_version": "2.0.0",  # Replace with your PyTorch Lightning version
    "hyper_parameters": {},  # Add any hyperparameters if needed
}

# Save the new checkpoint
torch.save(lightning_checkpoint, lightning_ckpt_path)

print(f"Converted checkpoint saved to: {lightning_ckpt_path}")
```
