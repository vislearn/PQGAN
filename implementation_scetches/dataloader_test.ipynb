```python
loader = LSUNLoader(resolution=256,
                    train_classes="train",
                    val_classes="val",
                    batch_size=8,
                    workers=8,
                    verbose=True,
                    data_dir="/path/to/lsun")
```

```python
loader = Places365Loader(resolution=256,
                         batch_size=8,
                         workers=8,
                         verbose=True,
                         data_dir="/path/to/places365")
```

```python
loader.prepare_data()
```

```python
loader.setup(stage="fit")
```

```python
dataloader = loader.train_dataloader()
```

```python
print(loader)
```

```python
import matplotlib.pyplot as plt
import torchvision.transforms as T

# Get a batch of data
data_iter = iter(dataloader)
images, labels = next(data_iter)

# Define a function to display images
def show_images(images, num_images=4):
    images = images[:num_images]  / 2 + 0.5  # Select a few images
    images = T.ToPILImage()(images[0]) if len(images.shape) == 3 else [T.ToPILImage()(img) for img in images]
    fig, axes = plt.subplots(1, num_images, figsize=(15, 5))
    for i, img in enumerate(images):
        axes[i].imshow(img)
        axes[i].axis('off')
    plt.show()

# Display the images
show_images(images)
```
