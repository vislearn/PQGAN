```python
import torchvision.models as models
import torch
import torch.nn as nn
```

```python
resnet18 = models.resnet18(pretrained=True)
resnet18.eval()
```

```python
test_model = nn.Sequential(*list(resnet18.children())[:-2])
```

```python
x = torch.randn(1, 3, 128, 128)

print(test_model(x).shape)
```

```python
upconv = nn.ConvTranspose2d(in_channels=6,
                            out_channels=3,
                            kernel_size=4,
                            stride=2,
                            padding=1,
                            output_padding=0,)

x = torch.randn(1, 6, 22,22)

print(upconv(x).shape)
```

```python
upsample = nn.Upsample(size=(22,22),
                                   mode='bilinear',
                                   align_corners=True)
x = torch.randn(5, 10, 11, 11)
print(upsample)
print(upsample(x).shape)
```

```python
conv = nn.Conv2d(in_channels=3,
                 out_channels=3,
                 kernel_size=2,
                 stride=2,
                 padding=3)

x = torch.randn(1, 3, 16, 16)

print(conv(x).shape)
```
