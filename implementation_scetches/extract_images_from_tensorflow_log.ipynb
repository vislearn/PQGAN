```python
import tensorflow as tf
import os
import numpy as np
from torchvision import transforms
from PIL import Image
import torch
```

```python
log_dir = "./"
tag = "original_vs_reconstruction"
event_file = os.path.join(log_dir, 'events.out.tfevents.1742661014.compgpu10.933882.0')  # The event file you want to inspect

# Normalize the tensor back to [0, 255] for saving as an image
normalize = transforms.Normalize(mean=[0, 0, 0], std=[1/0.5, 1/0.5, 1/0.5])

def tensor_to_pil(tensor):
    tensor = tensor.squeeze()  # Remove batch dimension if present
    tensor = tensor.numpy()
    tensor = np.clip(tensor * 255, 0, 255).astype(np.uint8)  # Convert to [0, 255]
            
    print(tensor.shape)
    return Image.fromarray(tensor)

for e in tf.compat.v1.train.summary_iterator(event_file):
    for v in e.summary.value:
        if v.tag == tag:
            # Get the image grid tensor (This is the 4D tensor [N, C, H, W])
            image_tensor = tf.image.decode_png(v.image.encoded_image_string)
            tf.keras.preprocessing.image.save_img(f"extracted_image_{e.step}.png", image_tensor.numpy())
```
